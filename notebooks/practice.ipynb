```python
import matplotlib.pyplot as plt
import numpy as np
```

### INDICACIONES

#### 1) Generar datos (distribución normal y uniforme)

```python
media_n = 200
desviacion = 25
muestra = 1000

distribucion_n = np.random.normal(media_n, desviacion, muestra)
distribucion_n
```

```python
limite_inferior = 50
limite_superior = 150
muestra = 1000

distribucion_u = np.random.uniform(limite_inferior, limite_superior, muestra)
distribucion_u
```

#### 2) Análisis estadísico de ambas distribuciones

```python
print('Distribución normal: ')
media = np.mean(distribucion_n)
mediana = np.median(distribucion_n)
desvio_est = np.std(distribucion_n)
varianza = np.var(distribucion_n)
minimo = np.min(distribucion_n)
maximo = np.max(distribucion_n)

print(f'La media es: {media}')
print(f'La mediana es: {mediana}')
print(f'El desvío estandar es: {desvio_est}')
print(f'La varianza es: {varianza}')
print(f'Minimo: {minimo}')
print(f'Maximo: {maximo}')
```

```python
print('Distribución uniforme: ')
media_u = np.mean(distribucion_u)
mediana_u = np.median(distribucion_u)
desvio_est_u = np.std(distribucion_u)
varianza_u = np.var(distribucion_u)
minimo_u = np.min(distribucion_u)
maximo_u = np.max(distribucion_u)

print(f'La media es: {media_u:.2f}')
print(f'La mediana es: {mediana_u:.2f}')
print(f'El desvío estandar es: {desvio_est_u:.2f}')
print(f'La varianza es: {varianza_u:.2f}')
print(f'Minimo: {minimo_u:.2f}')
print(f'Maximo: {maximo_u:.2f}')
```

#### 3) Visualización de ambas distribuciones

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 10))

ax1.hist(distribucion_n, bins='sqrt', color='lightblue', edgecolor='white')
ax1.set_title('Distribución normal')
ax1.set_xlabel('Valor')
ax1.set_ylabel('Frecuencia')
ax1.grid()

ax2.hist(distribucion_u, bins='sqrt', color='pink', edgecolor='white')
ax2.set_title('Distribución uniforme')
ax2.set_xlabel('Valor')
ax2.set_ylabel('Frecuencia')
ax2.grid()

fig.suptitle('Distribuciones')
plt.tight_layout()
plt.show()
```

#### 4) Reflexión

En el histograma, la distribución normal presenta una forma de campana, donde la mayoría de los valores se concentra alrededor de la media 198,95 y las frecuencias disminuyen hacia los extremos. Su media y mediana son casi iguales 198,95 y 199,02, y su desvío estándar es 24,23. En cambio, la distribución uniforme muestra frecuencias más parejas a lo largo del intervalo de 50 a 150; sus valores se centran cerca de 100 con media 100,54 y mediana 100,15. Aunque sus valores cubren un intervalo más acotado, en esta muestra la uniforme tiene mayor dispersión, con desvío estándar de 28,61 frente a 24,23 y varianza de 818,76 frente a 586,88.
